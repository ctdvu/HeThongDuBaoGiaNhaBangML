# Huấn luyện và so sánh mô hình dự đoán giá nhà Việt Nam

Bài toán có biến mục tiêu `Price` liên tục (tỷ VNĐ), nên notebook chỉ dùng các **mô hình hồi quy phù hợp**: Linear Regression, KNN Regressor, Decision Tree Regressor, Random Forest Regressor và Support Vector Regressor (SVR).

Các thuật toán phân loại hoặc không giám sát (Logistic Regression, Naive Bayes, KMeans, PCA) không được dùng vì không dự đoán trực tiếp giá nhà.

```python
from pathlib import Path
import warnings

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score
from sklearn.neighbors import KNeighborsRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import LinearSVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split

warnings.filterwarnings('ignore')
RANDOM_STATE = 42
DATA_PATH = Path('data/vietnam_housing_dataset.csv')
ARTIFACT_DIR = Path('artifacts')
ARTIFACT_DIR.mkdir(exist_ok=True)
sns.set_theme(style='whitegrid', palette='deep')
```

```python
# Đọc dữ liệu và tạo biến tỉnh/thành từ địa chỉ. Không dùng địa chỉ đầy đủ
# để tránh số lượng danh mục quá lớn và để form web dễ nhập liệu.
df = pd.read_csv(DATA_PATH)

def get_province(address):
    if pd.isna(address) or not str(address).strip():
        return 'Không xác định'
    province = str(address).rsplit(',', 1)[-1].strip()
    return province if province else 'Không xác định'

df['Province'] = df['Address'].apply(get_province)
target_column = 'Price'
numeric_features = ['Area', 'Frontage', 'Access Road', 'Floors', 'Bedrooms', 'Bathrooms']
categorical_features = ['House direction', 'Balcony direction', 'Legal status', 'Furniture state', 'Province']
feature_columns = numeric_features + categorical_features

for column in numeric_features + [target_column]:
    df[column] = pd.to_numeric(df[column], errors='coerce')
for column in categorical_features:
    df[column] = df[column].fillna('Không xác định').astype(str).str.strip().replace('', 'Không xác định')

# Bỏ các dòng không có giá/diện tích hợp lệ và các ngoại lệ cực đoan để đánh giá ổn định hơn.
model_df = df.dropna(subset=[target_column]).copy()
model_df = model_df[model_df['Area'].gt(0) & model_df[target_column].gt(0)]
low, high = model_df[target_column].quantile([0.005, 0.995])
model_df = model_df[model_df[target_column].between(low, high)].copy()

print(f'Dữ liệu ban đầu: {len(df):,} dòng')
print(f'Dữ liệu dùng huấn luyện: {len(model_df):,} dòng')
display(model_df[feature_columns + [target_column]].head())
```

```python
X = model_df[feature_columns].copy()
y = model_df[target_column].copy()
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE
)

def make_preprocessor():
    numeric_pipe = Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())
    ])
    categorical_pipe = Pipeline([
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('onehot', OneHotEncoder(handle_unknown='ignore'))
    ])
    return ColumnTransformer([
        ('numeric', numeric_pipe, numeric_features),
        ('categorical', categorical_pipe, categorical_features)
    ])

print(f'Train: {X_train.shape[0]:,} dòng | Test: {X_test.shape[0]:,} dòng')
```

```python
# Các mô hình đều nhận cùng tập train/test và cùng bước tiền xử lý để so sánh công bằng.
models = {
    'Linear Regression': LinearRegression(),
    'KNN Regressor': KNeighborsRegressor(n_neighbors=15, weights='distance', p=2),
    'Decision Tree': DecisionTreeRegressor(max_depth=18, min_samples_leaf=4, random_state=RANDOM_STATE),
    'Random Forest': RandomForestRegressor(
        n_estimators=100, max_depth=16, min_samples_leaf=5,
        random_state=RANDOM_STATE, n_jobs=-1
    ),
    'Linear SVR': LinearSVR(C=1.0, epsilon=0.15, max_iter=20000, dual='auto', random_state=RANDOM_STATE)
}

fitted_models, predictions, rows = {}, {}, []
for name, estimator in models.items():
    pipeline = Pipeline([('preprocessor', make_preprocessor()), ('model', estimator)])
    pipeline.fit(X_train, y_train)
    prediction = pipeline.predict(X_test)
    fitted_models[name] = pipeline
    predictions[name] = prediction
    rows.append({
        'Mô hình': name,
        'MAE (tỷ VNĐ)': mean_absolute_error(y_test, prediction),
        'RMSE (tỷ VNĐ)': root_mean_squared_error(y_test, prediction),
        'R²': r2_score(y_test, prediction)
    })

results = pd.DataFrame(rows).sort_values(['R²', 'MAE (tỷ VNĐ)'], ascending=[False, True]).reset_index(drop=True)
display(results.style.format({'MAE (tỷ VNĐ)': '{:.3f}', 'RMSE (tỷ VNĐ)': '{:.3f}', 'R²': '{:.3f}'}))
best_name = results.loc[0, 'Mô hình']
best_model = fitted_models[best_name]
print(f'Mô hình được chọn để triển khai: {best_name}')
```

```python
# Biểu đồ so sánh 5 thuật toán và chất lượng mô hình tốt nhất.
fig, axes = plt.subplots(1, 3, figsize=(20, 5.5))
plot_results = results.sort_values('R²')
sns.barplot(data=plot_results, x='R²', y='Mô hình', ax=axes[0], color='#176b87')
axes[0].set_title('So sánh R² (cao hơn tốt hơn)')
axes[0].set_xlim(min(-0.05, plot_results['R²'].min() - 0.05), 1)

plot_results = results.sort_values('MAE (tỷ VNĐ)', ascending=False)
sns.barplot(data=plot_results, x='MAE (tỷ VNĐ)', y='Mô hình', ax=axes[1], color='#f1a208')
axes[1].set_title('So sánh MAE (thấp hơn tốt hơn)')

axes[2].scatter(y_test, predictions[best_name], alpha=.35, s=16, color='#176b87')
min_value = min(y_test.min(), predictions[best_name].min())
max_value = max(y_test.max(), predictions[best_name].max())
axes[2].plot([min_value, max_value], [min_value, max_value], '--', color='#b42318')
axes[2].set_title(f'Giá thật và dự đoán: {best_name}')
axes[2].set_xlabel('Giá thật (tỷ VNĐ)')
axes[2].set_ylabel('Giá dự đoán (tỷ VNĐ)')

plt.tight_layout()
plt.savefig(ARTIFACT_DIR / 'model_comparison.png', dpi=160, bbox_inches='tight')
plt.show()
```

```python
# Lưu duy nhất mô hình thắng cùng metadata để app.py nạp trực tiếp.
category_values = {
    column: sorted(model_df[column].dropna().astype(str).unique().tolist())
    for column in categorical_features
}
best_metrics = results.iloc[0]
bundle = {
    'model': best_model,
    'model_name': best_name,
    'feature_columns': feature_columns,
    'category_values': category_values,
    'metrics': {
        'mae': float(best_metrics['MAE (tỷ VNĐ)']),
        'rmse': float(best_metrics['RMSE (tỷ VNĐ)']),
        'r2': float(best_metrics['R²'])
    },
    'target_unit': 'tỷ VNĐ'
}
model_path = ARTIFACT_DIR / 'housing_price_model.joblib'
joblib.dump(bundle, model_path)
print(f'Đã lưu mô hình: {model_path.resolve()}')

# Kiểm tra nhanh file vừa lưu.
loaded_bundle = joblib.load(model_path)
assert loaded_bundle['model_name'] == best_name
print('Kiểm tra lưu/tải mô hình thành công.')
```
